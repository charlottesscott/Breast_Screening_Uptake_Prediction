**Standardise Column Names**

```python
def clean_columns(df):
    df.columns = (
        df.columns
        .str.lower()
        .str.strip()
        .str.replace(" ", "_")
        .str.replace(r"[^\w]", "", regex=True)
    )
    return df
```

```python
uptake = clean_columns(uptake)
coverage = clean_columns(coverage)
cancers = clean_columns(cancers)
diagnostic = clean_columns(diagnostic)
referral = clean_columns(referral)
```

**Joining Keys**

```python
kc62_key = [
    "collectionyearrange",
    "parent_org_code",
    "org_name",
    "org_type",
    "age_band",
    "table_code"
]
```

```python
kc62_key = [
    "collectionyearrange",
    "parent_org_code", "parent_org_name",
    "org_name",
    "org_type",
    "age_band",
    "table_code"
]

for name, df in {
    "cancers": cancers,
    "diagnostic": diagnostic,
    "uptake": uptake,
    "referral": referral
}.items():

    duplicate_count = df.duplicated(
        subset=kc62_key,
        keep=False
    ).sum()

    print(name, duplicate_count)
```

```python
kc63_key = [
    "collectionyearrange",
    "parent_orgonscode",
    "parent_org_name",
    "org_name",
    "org_type",
    "age_band"
]

duplicate_count = coverage.duplicated(
    subset=kc63_key,
    keep=False
).sum()

print("coverage", duplicate_count)
```

```python
kc63_key2 = [
    "collectionyearrange",
    "parent_orgonscode",
    "parent_org_name",
    "org_onscode",
    "org_name",
    "org_type",
    "age_band"
]

duplicate_count = coverage.duplicated(
    subset=kc63_key,
    keep=False
).sum()

print("coverage", duplicate_count)
```

**Adding org code to KC62**

```python
org_lookup = coverage[
    ["org_onscode",
     "parent_orgonscode",
     "org_name",
     "org_type"]
].drop_duplicates()

org_lookup.groupby(
    ["org_name","org_type"]
)["org_onscode"].nunique().sort_values(ascending=False)
```

```python
coverage.loc[
    coverage["org_name"].isin([
        "Buckinghamshire",
        "Dorset",
        "North Yorkshire",
        "Somerset"
    ]),
    [
        "collectionyearrange",
        "parent_orgonscode",
        "org_onscode",
        "org_name",
        "org_type"
    ]
].drop_duplicates().sort_values(
    ["org_name", "collectionyearrange"]
)
```

**Validating that kc62_key is unique**

```python
for name, df in {
    "uptake": uptake,
    "cancers": cancers,
    "diagnostic": diagnostic,
    "referral": referral
}.items():

    dupes = df.duplicated(
        subset=kc62_key
    ).sum()

    print(name, dupes)
```

**Merge**

```python
master = uptake.merge(
    cancers,
    on=kc62_key,
    how="left"
)

master = master.merge(
    diagnostic,
    on=kc62_key,
    how="left"
)

master = master.merge(
    referral,
    on=kc62_key,
    how="left"
)
```
