**Import Python Libraries**

```python
import pandas as pd
import numpy as np

RANDOM_STATE = 42
```

**Create a reusable function**

```python
def load_csv(file_path, year):

    df = pd.read_csv(file_path)

    df["year"] = year

    return df
```

**Load the files using function**

Adding year into file

```python
kc63_2021 = load_csv(
    "kc63_coverage_21-22.csv",
    "2021-22"
)

kc63_2022 = load_csv(
    "kc63_coverage_22-23.csv",
    "2022-23"
)

kc63_2023 = load_csv(
    "kc63_coverage_23-24.csv",
    "2023-24"
)

kc63_2024 = load_csv(
    "kc63_coverage_24-25.csv",
    "2024-25"
)
```

**Combine all years**

Creating one large table

```python
kc63_all = pd.concat([
    kc63_2021,
    kc63_2022,
    kc63_2023,
    kc63_2024
])
```

**Standardise Column Names**

```python
def clean_columns(df):

    df.columns = (
        df.columns
        .str.lower()
        .str.strip()
        .str.replace(" ", "_")
    )

    return df
```

```python
kc63_all = clean_columns(kc63_all)
```

```python
kc63_all.to_csv(
    "kc63_all_clean.csv",
    index=False
)
```
