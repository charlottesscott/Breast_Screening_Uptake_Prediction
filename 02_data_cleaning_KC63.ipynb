**Create a reusable function**

```python
def load_csv(file_path, year):

    df = pd.read_csv(file_path)

    df["year"] = year

    return df
```

**Standardise Column Names**

```python
def clean_columns(df):

    df.columns = (
        df.columns
        .str.lower()
        .str.strip()
        .str.replace(" ", "_")
    )

    return df
```

```python
kc63_2024 = clean_columns(kc63_2024)
```

```python
kc63_2024.to_csv(
    "kc63_2024_clean.csv",
    index=False
)
```
